1.1	 Organização e Descrição da Telemetria

Os dados de telemetria representam os parâmetros vitais coletados pelos sensores da espaçonave antes do lançamento:

1.1.1 Temperatura Interna (°C) e Externa (°C): Monitoram o ambiente operacional dos equipamentos sensíveis e a estrutura externa.
Faixa segura interna: 15 °C a 28 °C.
Faixa segura externa: -40 °C a 60 °C (em plataforma)

As faixas operacionais de temperatura interna (15°C a 28°C) e externa (-40°C a 60°C) adotadas neste projeto seguem as diretrizes do General Environmental Verification Standard (GEVS) da NASA e a norma ECSS-E-ST-31C da Agência Espacial Europeia. Tais limites garantem a integridade dos subsistemas eletrônicos de bordo e a prevenção de choques térmicos na estrutura do veículo durante o abastecimento e a contagem regressiva na plataforma.

1.1.2 Integridade Estrutural (0 ou 1): Sinal binário vindo dos sensores de deformação e vibração (1 = Estrutura Integra/OK -  True ; 0 = Falha/Dano detectado - False).
1.1.3 Níveis de Energia (%): Percentual de carga acumulada nas baterias principais.
Nível mínimo seguro: ≥ 85%.
A exigência de carga mínima inicial da bateria de 85% garante a retenção da margem de contingência recomendada pela NASA-STD-4003, cobrindo o alto consumo dos atuadores durante a decolagem e mantendo a autonomia necessária em caso de reorientação orbital inicial

1.1.4 Pressão dos Tanques (PSI): Pressão dos propelentes e oxidantes nos tanques de combustível.
Faixa segura: 2.800 PSI a 3.200 PSI.

As faixas de tolerância estipuladas para a pressão dos tanques de propelente (2.800 a 3.200 PSI) têm como referência as diretrizes da norma ANSI/AIAA S-080 e da ISO 14623, que definem a Pressão Máxima de Operação Esperada (MEOP) e os fatores de segurança para evitar falhas estruturais.
1.1.5 Status dos Módulos Críticos: Leitura do estado operacional dos sistemas vitais (Propulsão, Suporte de Vida, Comunicação e Navegação). Todos devem retornar estado TRUE (OK).

1.3	 Script em Python


Este script simula a leitura dos dados de telemetria, executa as validações condicionais e exibe o resultado do checklist de pré-lançamento.


```python
# Criação das Variáveis de controle

temp_int_min = 15
temp_int_max = 28
temp_ext_min = -40
temp_ext_max = 60
estrutura_integral = True
nível_energia_min = 85
pressão_tanque_min = 2800
pressão_tanque_máx = 3200
mod_propulsao = True
mod_vida = True
mod_comunic = True
mod_navegacao = True

# integrar as variáveis no sistema, por digitação, ou leitura de equipamentos de forma automática

temp_int = float(input("Digite a temperatura interna em °C: "))
temp_ext = float(input("Digite a temperatura externa em °C: "))
pressão_tanque = float(input("Digite a pressão do tanque em psi: "))
nivel_energia = float(input("Digite o nível de energia em %: "))
estrutura_integral = int(input("Digite se a estrutura está íntegra (1) ou não(0): "))
mod_propulsao = int(input("Digite se o módulo de propulsão está ativo (1) ou não(0): "))
mod_vida = int(input("Digite se o módulo de vida está ativo (1) ou não(0): "))
mod_comunic = int(input("Digite se o módulo de comunicação está ativo (1) ou não(0): "))
mod_navegacao = int(input("Digite se o módulo de navegação está ativo (1) ou não(0): "))

# verificação do sistema com a função IF_ELSE

if temp_int >= temp_int_min and temp_int <= temp_int_max:
    texto_temp_int = "Temperatura interna OK"
    xti = 10
else:
    texto_temp_int = "Temperatura interna fora do padrão"
    xti = 0


if temp_ext >= temp_ext_min and temp_ext <= temp_ext_max:
    texto_temp_ext = "Temperatura externa OK"
    xte = 10
else:
    texto_temp_ext = "Temperatura externa fora do padrão"
    xte = 0

if pressão_tanque >= pressão_tanque_min and pressão_tanque <= pressão_tanque_máx:
    texto_pressão_tanque = "Pressão dos tanques OK"
    xpt = 10
else:
    texto_pressão_tanque = "Pressão dos tanques fora do padrão"
    xpt = 0

if nivel_energia >= nível_energia_min:
    texto_nivel_energia = "Nível de Energia OK"
    xne = 10
else:
    texto_nivel_energia = "Nível Energia fora do padrão"
    xne = 0

if estrutura_integral == 1:
    texto_estrutura_integral = "Estrutura OK"
    xei = 10
else:
    texto_estrutura_integral = "Estrutura fora do padrão"
    xei = 0

if mod_propulsao == 1:
    texto_mod_propulsao = "Módulo de propulsão OK"
    xmp = 10
else:
    texto_mod_propulsao = "Módulo de propulção fora do padrão"
    xmp = 0

if mod_vida == 1:
    texto_mod_vida = "Módulo de vida OK"
    xmv = 10
else:
    texto_mod_vida = "Módulo de vida fora do padrão"
    xmv = 0

if mod_comunic == 1:
    texto_mod_comunic = "Módulo de comunicação OK"
    xmc = 10
else:
    texto_mod_comunic = "Módulo de comunicação fora do padrão"
    xmc = 0

if mod_navegacao == 1:
      texto_mod_navegacao = "Módulo de navegação OK"
      xmn = 10
else:
      texto_mod_navegacao = "Módulo de navegação fora do padrão"
      xmn = 0

# Impressão de cada variável se está dentro ou fora do controle
print ("="*50)
print()
print (f"{'LEITURA DA VARIÁVEIS':^50}")
print()
print ("="*50)
print()
print(texto_temp_int)
print(texto_temp_ext)
print(texto_pressão_tanque)
print(texto_nivel_energia)
print(texto_estrutura_integral)
print(texto_mod_propulsao)
print(texto_mod_vida)
print(texto_mod_comunic)
print(texto_mod_navegacao)
print()
print ("="*50)
print ()
# verificação de cada variável e da decolagem

verificacao = (xti + xte + xpt + xne + xei + xmp + xmv + xmc + xmn)

if verificacao == 90:
  print (f"{'DECOLAGEM PERMITIDA':^50}")
else:
  print (f"{'DECOLAGEM NÃO PERMITIDA':^50}")
print()
print("="*50)

# FIM
```
